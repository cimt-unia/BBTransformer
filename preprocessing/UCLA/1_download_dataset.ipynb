```python
from nilearn import datasets
import os
import re

# 1. Setup Directory
target_dir = "/mnt/movement/users/jaizor/xtra/data/fmri"
os.makedirs(target_dir, exist_ok=True)

print(f"🦖 Iku says: Checking available URLs for ds000030...")

# 2. Fetch URLs (Let Nilearn decide the stable version it knows)
# Note: We do NOT pass a version arg here because fetch_ds000030_urls 
# typically doesn't support arbitrary live tags like '00016' reliably.
urls_path, all_urls = datasets.fetch_ds000030_urls(data_dir=target_dir, verbose=1)

# Inspect the first URL to see what version Nilearn actually gave us
if all_urls:
    detected_version = all_urls[0].split('/')[3] # e.g., ds000030_R1.0.4
    print(f"✅ Nilearn provided URLs for version: {detected_version}")
else:
    raise RuntimeError("No URLs fetched. Check internet connection.")

# 3. Filter for fMRI only (Save disk space!)
fmri_pattern = re.compile(r".*(_bold\.nii\.gz|_bold\.json|_events\.tsv)$")
fmri_urls = [u for u in all_urls if fmri_pattern.match(u)]

print(f"📉 Filtering complete: {len(fmri_urls)} fMRI files selected out of {len(all_urls)} total.")

# 4. Download using the DETECTED version string
# This prevents the "prefix not found" error because we match the URL content.
print(f"🚀 Starting download for {detected_version}...")
try:
    data_dir, files = datasets.fetch_openneuro_dataset(
        urls=fmri_urls,
        dataset_version=detected_version, # CRITICAL: Use the version we just detected!
        data_dir=target_dir,
        verbose=1
    )
    print(f"🎉 Success! Data located at: {data_dir}")
except Exception as e:
    print(f"❌ Download failed: {e}")
    print("💡 Colleague advice: If this fails, your Nilearn version might be too old to handle the S3 redirect correctly. Consider using 'datalad' via CLI instead.")
```
